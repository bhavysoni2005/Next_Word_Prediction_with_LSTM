```python
from tensorflow.keras.preprocessing.text import Tokenizer
```

```python
vocab_size = 10000
tokinizer= Tokenizer(num_words=vocab_size,oov_token="<oov>")
tokinizer.fit_on_texts(quotes)
```

```python
from IPython.utils.tempdir import TemporaryWorkingDirectory
word_index = tokinizer.word_index
print(len(word_index))
list(word_index.items())[:10]
```

```python
from typing import Sequence
Sequence = tokinizer.texts_to_sequences(quotes)
for i in range(5):
  print(quotes[i])
  print(Sequence[i])
```

```python
X = []
y = []
for i in range(len(Sequence)):
  for j in range(1,len(Sequence[i])):
    X.append(Sequence[i][:j])
    y.append(Sequence[i][j])
```

```python
max_len = max(len(x) for x in X)
max_len 
```

```python
# Now apply padding for equilizing the input and output 
from tensorflow.keras.preprocessing.sequence import pad_sequences
X_padded = pad_sequences(X,maxlen=max_len,padding='pre')
```

```python
X_padded[0]
```

```python
y = np.array(y)
```

```python
#One Hot encode the y variable
from tensorflow.keras.utils import to_categorical
y_one_hot = to_categorical(y,num_classes=vocab_size)
```

```python
print(y_one_hot.shape)
```

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding,SimpleRNN,LSTM,Dense
```

```python
embedding_dim = 50
rnn_unit = 128
```

```python
rnn_model = Sequential()
rnn_model.add(Embedding(input_dim=vocab_size,output_dim=embedding_dim,input_length=max_len))
rnn_model.add(SimpleRNN(units = rnn_unit))
rnn_model.add(Dense(units = vocab_size,activation='softmax')) 
```

```python
rnn_model.compile(
    optimizer = 'adam',
    loss = 'categorical_crossentropy',
    metrics = ['accuracy']
)
```

```python
rnn_model.summary()
```

```python
#LSTM Model
lstm_model = Sequential()
lstm_model.add(Embedding(input_dim=vocab_size,output_dim=embedding_dim,input_length = max_len))
lstm_model.add(LSTM(units = rnn_unit))
lstm_model.add(Dense(units = vocab_size,activation='softmax'))
```

```python
lstm_model.compile(
    optimizer = 'adam',
    loss = 'categorical_crossentropy',
    metrics = ['accuracy']
)
```

```python
lstm_model.summary()
```

```python
epochs = 10
batch_size = 128
```

```python
history_rnn = rnn_model.fit(
    X_padded,
    y_one_hot,
    epochs = epochs,
    batch_size = batch_size,
    validation_split = 0.1
)
```

```python
index_to_word = {}
for word, index in word_index.items():
  index_to_word[index] = word
     
```

```python
from tensorflow.keras.preprocessing.sequence import pad_sequences
     
```

```python
def predictor(model,tokenizer,text,max_len):
  text = text.lower()

  seq = tokenizer.texts_to_sequences([text])[0]
  seq = pad_sequences([seq], maxlen=max_len, padding='pre')

  pred = model.predict(seq,verbose = 0)
  pred_index = np.argmax(pred)
  return index_to_word[pred_index]
```

```python
seed_text = "what are you"
next_word = predictor(lstm_model,tokinizer,seed_text,max_len)
print(next_word)
```

```python
def generate_text(model,tokenizer,seed_text,max_len,n_words):
  for _ in range(n_words):
    next_word = predictor(model,tokenizer,seed_text,max_len)
    if next_word == "":
      break
    seed_text += " " + next_word
  return seed_text
```

```python
seed = "are you a "
generate_text = generate_text(lstm_model,tokinizer,seed,max_len,10)
print(generate_text)
```

```python
# Advanced Text Generation with Better Repetition Prevention
def generate_text_advanced(model, tokenizer, seed_text, max_len, n_words, 
                           temperature=1.0, prevent_repetition=2, diversity_penalty=0.0):
    """
    Advanced text generation with multiple strategies to avoid repetition.
    
    Args:
        temperature: Higher values increase diversity
        prevent_repetition: Number of previous words to prevent repeating
        diversity_penalty: Penalize repeated n-grams (0 to 1)
    """
    generated_words = seed_text.split()
    
    for iteration in range(n_words):
        # Get prediction
        seq = tokenizer.texts_to_sequences([seed_text])[0]
        seq = pad_sequences([seq], maxlen=max_len, padding='pre')
        
        preds = model.predict(seq, verbose=0)[0].copy()
        
        # Apply diversity penalty to recent words
        if diversity_penalty > 0:
            for recent_word in generated_words[-prevent_repetition:]:
                if recent_word in tokenizer.word_index:
                    idx = tokenizer.word_index[recent_word]
                    if idx < len(preds):
                        preds[idx] = preds[idx] * (1 - diversity_penalty)
        
        # Apply temperature
        if temperature != 1.0:
            preds = np.power(preds, 1.0/temperature)
            preds = preds / (np.sum(preds) + 1e-10)
            pred_index = np.random.choice(len(preds), p=preds)
        else:
            pred_index = np.argmax(preds)
        
        # Get word
        next_word = ""
        for word, index in tokenizer.word_index.items():
            if index == pred_index:
                next_word = word
                break
        
        if next_word == "":
            break
        
        generated_words.append(next_word)
        seed_text += " " + next_word
    
    return " ".join(generated_words)

# Compare different strategies
print("\n" + "=" * 70)
print("COMPARING DIFFERENT TEXT GENERATION STRATEGIES")
print("=" * 70)

seed = "life is"
n_words = 15

strategies = [
    ("Standard (temp=1.0)", 1.0, 0, 0.0),
    ("Conservative (temp=0.5)", 0.5, 0, 0.0),
    ("Diverse (temp=1.5)", 1.5, 0, 0.0),
    ("With Repetition Prevention", 1.0, 2, 0.0),
    ("High Diversity Penalty", 1.0, 2, 0.5),
    ("Conservative + Penalty", 0.7, 2, 0.3),
]

for name, temp, prevent, penalty in strategies:
    result = generate_text_advanced(lstm_model, tokinizer, seed, max_len, n_words,
                                    temperature=temp, prevent_repetition=prevent,
                                    diversity_penalty=penalty)
    print(f"\n{name}:")
    print(f"  Result: {result}")
    print("-" * 70)
```
